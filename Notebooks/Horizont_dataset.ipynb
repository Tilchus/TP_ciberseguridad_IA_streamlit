## Paso 1: Carga de Datos e Inspección Inicial

```python
import pandas as pd

# Rutas de tus dos archivos Parquet (ajustá los nombres si es necesario)
ruta_parquet_1 = '../datos/datos_sin_procesar/synthetic_direct_test.parquet'  # Reemplaza con el nombre de tu primer parquet
ruta_parquet_2 = '../datos/datos_sin_procesar/synthetic_docs_test.parquet'  # Reemplaza con el nombre de tu segundo parquet

# ==========================================
# 1. INSPECCIÓN DEL PRIMER ARCHIVO PARQUET
# ==========================================
print("📂 CARGANDO Y ANALIZANDO: Dataset 1")
df_1 = pd.read_parquet(ruta_parquet_1)

print("\n--- 1. Primeras 5 filas (para ver los datos) ---")
display(df_1.head())

print("\n--- 2. Información estructural (df.info()) ---")
df_1.info()

print("\n" + "="*70 + "\n")

# ==========================================
# 2. INSPECCIÓN DEL SEGUNDO ARCHIVO PARQUET
# ==========================================
print("📂 CARGANDO Y ANALIZANDO: Dataset 2")
df_2 = pd.read_parquet(ruta_parquet_2)

print("\n--- 1. Primeras 5 filas (para ver los datos) ---")
display(df_2.head())

print("\n--- 2. Información estructural (df.info()) ---")
df_2.info()
```

### 📊 Resumen Estructural de los Datos (Radiografía en Tabla)

* **¿Qué hicimos?** Creamos un script personalizado para transformar el reporte técnico de texto plano (`df.info()`) en un DataFrame limpio y estructurado.
* **¿Qué hace?** Construye y muestra una tabla interactiva detallando:
  - El **nombre** de cada columna.
  - El **tipo de dato** (`dtype`) que almacena.
  - La cantidad de **valores válidos** (no nulos) por campo.

```python
# Creamos un DataFrame resumen con la estructura del dataset
df_resumen = pd.DataFrame({
    'Columna': df_1.columns,
    'Tipo de Dato': df_1.dtypes.values,
    'Valores No Nulos': df_1.notnull().sum().values,
    'Valores Nulos': df_1.isnull().sum().values
})

# Lo mostramos como una tabla bonita con display()
display(df_resumen)
```

### 📊 Tabla Resumen de Distribución (Frecuencias y Porcentajes)
* **¿Qué hacemos?** Construimos un DataFrame estructurado para la columna `kind` (tipos de ataques) y la columna `lang` (idiomas/países registrados).
* **¿Qué calcula?** Muestra en formato de tabla limpia la **cantidad absoluta** de registros y su respectivo **porcentaje sobre el total** para evaluar proporciones de manera visual.

```python
import pandas as pd

# Función para armar una tabla bonita con cantidad y porcentaje
def generar_tabla_resumen(df, columna):
    # Calculamos conteo y porcentaje
    conteo = df[columna].value_counts()
    porcentaje = (df[columna].value_counts(normalize=True) * 100).round(2)
    
    # Unimos ambos resultados en un nuevo DataFrame limpio
    df_resultado = pd.DataFrame({
        'Cantidad': conteo,
        'Porcentaje (%)': porcentaje
    })
    
    return df_resultado

# 1. Tabla resumen para la columna 'kind' en df_1
print("--- 1. Distribución de Tipos (kind) en df_1 ---")
display(generar_tabla_resumen(df_1, 'kind'))

print("\n" + "="*50 + "\n")

# 2. Tabla resumen para la columna 'lang' (idiomas/países) en df_1
print("--- 2. Distribución por Idioma/País (lang) en df_1 ---")
display(generar_tabla_resumen(df_1, 'lang'))
```

### 🌐 Matriz de Frecuencias y Porcentajes Globales (Idioma vs. Tipo de Ataque)

Este bloque consolida el análisis integrando ambos datasets de Horizon (`df_1` y `df_2`) en un único corpus unificado para obtener una visión global de los datos sin sesgos individuales. 

Se presentan dos perspectivas analíticas complementarias:
1. **Matriz de Frecuencias (Conteo Absoluto):** Muestra la cantidad pura de registros distribuidos entre el idioma (`lang`) y el tipo de inyección (`kind`), incluyendo los totales marginales.
2. **Matriz de Porcentajes sobre el Gran Total (Opción A):** Calcula de forma automatizada qué proporción exacta representa cada celda en relación con el 100% de todo el corpus combinado (`normalize='all'`), facilitando la lectura ponderada de las frecuencias.

```python
import pandas as pd

# 1. Unificamos ambos datasets de Horizon en un único corpus global
df_horizon_total = pd.concat([df_1, df_2], ignore_index=True)

print(f"Total combinado del corpus Horizon: {len(df_horizon_total)} registros.")

# 2. Tabla 1: Conteo Absoluto (Cantidad pura de registros por Idioma vs Tipo de Ataque)
tabla_conteos = pd.crosstab(
    df_horizon_total['lang'], 
    df_horizon_total['kind'], 
    margins=True, 
    margins_name='Total General'
)

print("\n--- 1. MATRIZ DE FRECUENCIAS (Conteo Absoluto) ---")
display(tabla_conteos)

print("\n" + "="*60 + "\n")

# 3. Tabla 2 (Opción A): Porcentajes respecto al Gran Total 
# (Calcula exactamente qué porcentaje representa cada valor sobre el total general del dataset)
tabla_porcentajes = (pd.crosstab(
    df_horizon_total['lang'], 
    df_horizon_total['kind'], 
    normalize='all', 
    margins=True, 
    margins_name='Total General'
) * 100).round(2)

print("--- 2. MATRIZ DE PORCENTAJES (%) SOBRE EL GRAN TOTAL ---")
display(tabla_porcentajes)
```

### 📊 Matriz de Frecuencias Consolidada con Porcentaje del Gran Total

Este bloque procesa el corpus unificado (`df_1` y `df_2`) para generar una matriz de frecuencias avanzada que cruza los idiomas (`lang`) con los tipos de inyección (`kind`), incorporando métricas ponderadas clave:

* **Suma por Fila (`Total General`):** Agrega una columna que consolida el total de registros (ataques directos e indirectos sumados) correspondientes a cada idioma o país.
* **Porcentaje del Gran Total (`% del Total General`):** Calcula de forma automatizada y redondeada a dos decimales qué proporción exacta representa el total de cada fila en relación con el universo completo del dataset (el 100% de los registros globales).
* **Fila de Cierre (`Total General (Dataset)`):** Añade una fila final que totaliza las columnas numéricas y fija el 100.0% como referencia de control del corpus total.

**Objetivo Analítico:** Permitir visualizar de manera clara y directa qué volumen y qué incidencia porcentual global aporta cada idioma o región dentro del total de la muestra de seguridad.

```python
import pandas as pd

# 1. Unificamos ambos DataFrames en el corpus global
df_horizon_total = pd.concat([df_1, df_2], ignore_index=True)

# 2. Creamos la tabla cruzada base (directos e indirectos por idioma)
tabla = pd.crosstab(df_horizon_total['lang'], df_horizon_total['kind'])

# 3. Agregamos la columna de Total por fila (la suma de directos + indirectos de cada país)
tabla['Total General'] = tabla.sum(axis=1)

# 4. Calculamos el Gran Total de todo el dataset (los 6515 registros)
gran_total = tabla['Total General'].sum()

# 5. Agregamos la nueva columna con el porcentaje que representa cada fila respecto al Gran Total
tabla['% del Total General'] = ((tabla['Total General'] / gran_total) * 100).round(2)

# 6. Agregamos una fila al final con la suma total de todo
fila_totales = tabla.sum(numeric_only=True)
fila_totales.name = 'Total General (Dataset)'
tabla.loc['Total General (Dataset)'] = fila_totales

# Aseguramos que el porcentaje de la fila final sea 100%
tabla.loc['Total General (Dataset)', '% del Total General'] = 100.0

print("--- Matriz de Frecuencias con Columna de Porcentaje del Total General ---")
display(tabla)
```
